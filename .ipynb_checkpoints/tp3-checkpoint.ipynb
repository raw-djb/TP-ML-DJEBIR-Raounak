```python
LOO = LeaveOneOut()
```

```python
# MSE
def rmse(y_true, y_pred):
    return float(np.sqrt(mean_squared_error(y_true, y_pred)))
```

```python
def train_metrics(model):
    # for the all model 
    model.fit(X, y)
    pred = model.predict(X) 
    return rmse(y, pred), r2_score(y, pred) # Return prediction error and R²
```
